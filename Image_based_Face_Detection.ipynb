```python
import cv2
from google.colab.patches import cv2_imshow

front_face_detector = cv2.CascadeClassifier(
    cv2.data.haarcascades + "haarcascade_frontalface_alt.xml"
)

image = cv2.imread("/content/people.jpg")

grey_image = cv2.cvtColor(image, cv2.COLOR_BGR2GRAY)

face_detector = front_face_detector.detectMultiScale(grey_image)

print(face_detector)

for x, y, w, h in face_detector:
    cv2.rectangle(grey_image, (x, y), (x+w, y+h), (255, 0, 0), 3)

cv2_imshow(grey_image)
cv2.waitKey(0)
```
